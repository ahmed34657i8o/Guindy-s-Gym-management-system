```python
from abc import ABC, abstractmethod

class Person(ABC):
    def __init__(self, name, phone_num, person_id):
        self.name = name
        self.phone_num = phone_num
        self.person_id = person_id

    @property
    def name(self):
        return self.__name

    @name.setter
    def name(self, name):
        if not name:
            raise ValueError("Name cannot be empty")
        else:
            self.__name = name

    @property
    def phone_num(self):
        return self.__phone_num

    @phone_num.setter
    def phone_num(self, phone_num):
        if not phone_num.isdigit():
            raise ValueError("Phone number must be numbers only")
        else:
            self.__phone_num = phone_num

    @property
    def person_id(self):
        return self.__person_id

    @person_id.setter
    def person_id(self, person_id):
        if not person_id.isdigit():
            raise ValueError("ID must be numbers only")
        else:
            self.__person_id = person_id

    @abstractmethod
    def get_details(self):                  # every child must write its own method
        pass

    def __str__(self):
        return f"Name: {self.name} | ID: {self.person_id} | Phone: {self.phone_num}"
```

```python
from datetime import datetime

class Membership:
    def __init__(self, membership_type, start_date, end_date):
        self.membership_type = membership_type
        self.start_date = start_date
        self.end_date = end_date

    @property
    def membership_type(self):
        return self.__membership_type

    @membership_type.setter
    def membership_type(self, new_type):
        valid_types = ["Monthly", "Yearly", "Weekly"]
        if new_type not in valid_types:
            raise ValueError(f"Membership type must be one of {valid_types}")
        self.__membership_type = new_type

    @property
    def start_date(self):
        return self.__start_date

    @start_date.setter
    def start_date(self, new_start_date):
        self.__start_date = datetime.strptime(new_start_date, "%Y-%m-%d").date()

    @property
    def end_date(self):
        return self.__end_date

    @end_date.setter
    def end_date(self, new_end_date):
        self.__end_date = datetime.strptime(new_end_date, "%Y-%m-%d").date()

    def renew(self,new_end_date):
      self.start_date = self.end_date
      self.end_date = new_end_date


    def is_active(self):
     today = datetime.today().date()
     if self.start_date <= today <= self.end_date:
        return True
     else:
        return False

    def days_left(self):
     today = datetime.today().date()
     days_left = (self.end_date - today).days
     if days_left <= 0:
        return "Membership expired you dont have any days left"
     else:
      return days_left

    def __str__(self):
     if self.is_active():
        status = "Active"
     else:
        status = "Expired"

     return (f"Membership: {self.membership_type} , Start: {self.start_date} , End: {self.end_date} , Status: {status} , Days left: {self.days_left()}")


s = Membership('Yearly','2025-10-1','2026-08-01')
print(s)
```

```python

class Trainer(Person):
    def __init__(self, name, phone_num, person_id, specialty):
        super().__init__(name, phone_num, person_id)
        self.specialty = specialty
        self.__members = []

    @property
    def specialty(self):
        return self.__specialty

    @specialty.setter
    def specialty(self, specialty):
        if specialty is None:
            raise ValueError("Specialty cannot be empty")
        else:
         self.__specialty = specialty

    def add_member(self, member):
        if member not in self.__members:
            self.__members.append(member)

    def remove_member(self, member):
        if member not in self.__members:
            return False
        if member.membership is not None and not member.membership.is_active():
            self.__members.remove(member)
            return True
        return False

    def get_members(self):
        return self.__members

    def get_details(self):
        return {
            "name": self.name,
            "id": self.person_id,
            "phone": self.phone_num,
            "specialty": self.specialty,
            "members": len(self.__members)
        }


    def __str__(self):
        return f"Trainer: {self.name} | ID: {self.person_id} | Phone: {self.phone_num} | Specialty: {self.specialty} | Members: {len(self.__members)}"
```

```python

class Payment:
    def __init__(self, payment_id, member, amount, date):
        self.payment_id = payment_id
        self.member = member
        self.amount = amount
        self.date = date

    @property
    def payment_id(self):
        return self.__payment_id

    @payment_id.setter
    def payment_id(self, payement_id):
        if not payement_id.isdigit():
            raise ValueError("ID must be numbers only")
        else:
            self.__payment_id = payement_id

    @property
    def member(self):
        return self.__member

    @member.setter
    def member(self, member):
        if member is None:
            raise ValueError("Payment must belong to a member")
        else:
            self.__member = member

    @property
    def amount(self):
        return self.__amount

    @amount.setter
    def amount(self, amount):
      if type(amount) != int and type(amount) != float or amount <= 0:
        raise ValueError("Amount must be a number and bigger than 0")
      else:
        self.__amount = amount

    @property
    def date(self):
        return self.__date

    @date.setter
    def date(self, date):
        self.__date = datetime.strptime(date, "%Y-%m-%d").date()

    def get_details(self):
        return {
            "id": self.payment_id,
            "member": self.member.name,
            "amount": self.amount,
            "date": self.date   }

    def __str__(self):
        return (f"Payment: {self.payment_id} , Member: {self.member.name} , Amount: {self.amount} , Date: {self.date}")


p = Payment('1',ahmed,600,'2023-10-1')
print(p.get_details())
print(p)

```

```python
class Gym:
    def __init__(self, name: str):
        self.name = name
        self.members = []
        self.trainers = []
        self.payments = []
        self.attendances = []

    #  HELPER LOOKUPS
    def check_member_validity(self, member_id: str):
        """Returns member object if found, otherwise None."""
        search_id = str(member_id).strip()
        for m in self.members:
            # Check member_id property or fall back to person_id
            m_id = getattr(m, 'member_id', getattr(m, 'person_id', None))
            if str(m_id).strip() == search_id:
                return m
        return None

    def check_trainer_validity(self, trainer_id: str):
        """Returns trainer object if found, otherwise None."""
        search_id = str(trainer_id).strip()
        for t in self.trainers:
            # Check trainer_id property or fall back to person_id
            t_id = getattr(t, 'trainer_id', getattr(t, 'person_id', None))
            if str(t_id).strip() == search_id:
                return t
        return None

    #  FEATURE METHODS
    def add_member(self):
        try:
            raw_input = input("Enter Member ID, Name, Phone (space separated): ").strip()
            parts = raw_input.split()
            if len(parts) < 3:
                print("Error: Expected 3 values (ID, Name, Phone). Please try again.")
                return False

            member_id, name, phone = parts[0], parts[1], parts[2]

            if self.check_member_validity(member_id):
                print(f"Error: Member ID '{member_id}' already exists!")
                return False

            new_member = Member(name, phone, member_id)
            self.members.append(new_member)
            print(f"Member '{name}' added successfully!")
            return True
        except Exception as e:
            print(f"Input Error: {e}")
            return False

    def add_trainer(self):
        try:
            raw_input = input("Enter Trainer ID, Name, Phone, Specialty (space separated): ").strip()
            parts = raw_input.split()
            if len(parts) < 4:
                print("Error: Expected 4 values (ID, Name, Phone, Specialty). Please try again.")
                return False

            trainer_id, name, phone, specialty = parts[0], parts[1], parts[2], " ".join(parts[3:])

            if self.check_trainer_validity(trainer_id):
                print(f"Error: Trainer ID '{trainer_id}' already exists!")
                return False

            new_trainer = Trainer(name, phone, trainer_id, specialty)
            self.trainers.append(new_trainer)
            print(f"Trainer '{name}' added successfully!")
            return True
        except Exception as e:
            print(f"Input Error: {e}")
            return False

    def create_membership(self):
        try:
            raw_input = input("Enter Member ID, Type (Weekly/Monthly/Yearly), Start Date (YYYY-MM-DD), End Date (YYYY-MM-DD): ").strip()
            parts = raw_input.split()
            if len(parts) < 4:
                print("Error: Expected 4 values (Member ID, Type, Start Date, End Date). Please try again.")
                return False

            member_id, m_type, start_date, end_date = parts[0], parts[1], parts[2], parts[3]
            member = self.check_member_validity(member_id)

            if not member:
                print(f"Error: Member with ID '{member_id}' not found.")
                return False

            membership = Membership(m_type, start_date, end_date)
            member.set_membership(membership)
            print(f"Membership '{m_type}' assigned to {member.name} successfully!")
            return True
        except Exception as e:
            print(f"Error assigning membership: {e}")
            return False

    def assign_trainer(self):
        try:
            raw_input = input("Enter Member ID and Trainer ID (space separated): ").strip()
            parts = raw_input.split()
            if len(parts) < 2:
                print("Error: Expected 2 values (Member ID and Trainer ID). Please try again.")
                return False

            member_id, trainer_id = parts[0], parts[1]
            member = self.check_member_validity(member_id)
            trainer = self.check_trainer_validity(trainer_id)

            if not member:
                print(f"Error: Member ID '{member_id}' not found.")
                return False
            if not trainer:
                print(f"Error: Trainer ID '{trainer_id}' not found.")
                return False

            member.set_trainer(trainer)
            trainer.add_member(member)
            print(f"Trainer '{trainer.name}' assigned to Member '{member.name}'.")
            return True
        except Exception as e:
            print(f"Error assigning trainer: {e}")
            return False

    def record_payment(self):
        try:
            raw_input = input("Enter Payment ID, Member ID, Amount, Date (YYYY-MM-DD): ").strip()
            parts = raw_input.split()
            if len(parts) < 4:
                print("Error: Expected 4 values (Payment ID, Member ID, Amount, Date). Please try again.")
                return False

            payment_id, member_id, amount_str, date_str = parts[0], parts[1], parts[2], parts[3]
            member = self.check_member_validity(member_id)

            if not member:
                print(f"Error: Member ID '{member_id}' not found.")
                return False

            payment = Payment(payment_id, member, float(amount_str), date_str)
            self.payments.append(payment)
            print(f"Payment of ${payment.amount:.2f} recorded for {member.name}.")
            return True
        except Exception as e:
            print(f"Error recording payment: {e}")
            return False

    def record_attendance(self):
        try:
            raw_input = input("Enter Member ID and Date (YYYY-MM-DD): ").strip()
            parts = raw_input.split()
            if len(parts) < 2:
                print("Error: Expected 2 values (Member ID, Date). Please try again.")
                return False

            member_id, date_str = parts[0], parts[1]
            member = self.check_member_validity(member_id)

            if not member:
                print(f"Error: Member ID '{member_id}' not found.")
                return False

            attendance = Attendance(member, date_str)
            self.attendances.append(attendance)
            print(f"Attendance recorded for '{member.name}' on {date_str}.")
            return True
        except Exception as e:
            print(f"Attendance Error: {e}")
            return False

    def search_members_by_prefix(self):
        prefix = input("Enter member name prefix: ").strip()
        matches = [m for m in self.members if m.name.lower().startswith(prefix.lower())]

        if not matches:
            print("No members found with that prefix.")
            return False

        print("\n--- Search Results ---")
        for m in matches:
            print(m)
        return True

    def print_active_memberships(self):
        active_list = [m for m in self.members if getattr(m, 'membership', None) and m.membership.is_active()]
        if not active_list:
            print("No active memberships found.")
            return

        print("\n--- Active Memberships ---")
        for m in active_list:
            print(f"{m} | {m.membership}")

    def print_expired_memberships(self):
        expired_list = [m for m in self.members if not getattr(m, 'membership', None) or not m.membership.is_active()]
        if not expired_list:
            print("No expired/inactive memberships found.")
            return

        print("\n--- Expired / Inactive Memberships ---")
        for m in expired_list:
            status_str = m.membership if getattr(m, 'membership', None) else "No Membership Created"
            print(f"{m} | Status: {status_str}")

    def display_dashboard(self):
        total_members = len(self.members)
        active_memberships = len([m for m in self.members if getattr(m, 'membership', None) and m.membership.is_active()])
        expired_memberships = total_members - active_memberships
        total_revenue = sum(p.amount for p in self.payments)

        print(f"\n======== {self.name.upper()} DASHBOARD ========")
        print(f"Total Members:             {total_members}")
        print(f"Active Memberships:        {active_memberships}")
        print(f"Expired/No Memberships:    {expired_memberships}")
        print(f"Total Trainers:            {len(self.trainers)}")
        print(f"Total Payments Recorded:   {len(self.payments)}")
        print(f"Total Revenue Generated:   ${total_revenue:.2f}")
        print(f"Total Check-in Logs:       {len(self.attendances)}")
        print("==========================================")

    #  MENU SYSTEM
    def menu(self):
        choice = -1
        while choice == -1:
            print(f"\n--- {self.name} Management Menu ---")
            print("1) Add Member")
            print("2) Add Trainer")
            print("3) Create/Assign Membership")
            print("4) Assign Trainer to Member")
            print("5) Record Attendance")
            print("6) Record Payment")
            print("7) Search Members by Prefix")
            print("8) Display Active Memberships")
            print("9) Display Expired Memberships")
            print("10) Show Dashboard Stats")
            print("11) Exit")

            raw = input("Choose an option (1-11): ").strip()
            if raw.isdigit() and 1 <= int(raw) <= 11:
                choice = int(raw)
            else:
                print("Invalid choice. Please enter a number between 1 and 11.")
                choice = -1
        return choice

    def run_system(self):
        actions = {
            1: self.add_member,
            2: self.add_trainer,
            3: self.create_membership,
            4: self.assign_trainer,
            5: self.record_attendance,
            6: self.record_payment,
            7: self.search_members_by_prefix,
            8: self.print_active_memberships,
            9: self.print_expired_memberships,
            10: self.display_dashboard,
        }
        while True:
            choice = self.menu()
            if choice == 11:
                print("\nExiting Guindy's Gym Management System. Goodbye!")
                break
            actions[choice]()



# SYSTEM EXECUTION

if __name__ == "__main__":
    gym = Gym("Guindy's Fitness Gym")
    gym.run_system()
```

```python
# ==========================================
# 1. OOP CORE CLASSES WITH DICT CONVERSIONS
# ==========================================

import json
import os
from datetime import datetime

class Person:
    def __init__(self, name: str, phone: str, person_id: str):
        self.name = name
        self.phone = phone
        self.person_id = person_id

    def __str__(self):
        return f"ID: {self.person_id} | Name: {self.name} | Phone: {self.phone}"


class Membership:
    def __init__(self, m_type: str, start_date: str, end_date: str):
        self.m_type = m_type
        self.start_date = start_date
        self.end_date = end_date

    def is_active(self) -> bool:
        try:
            today = datetime.now().date()
            end = datetime.strptime(self.end_date, "%Y-%m-%d").date()
            return today <= end
        except ValueError:
            return True

    def to_dict(self):
        return {
            "m_type": self.m_type,
            "start_date": self.start_date,
            "end_date": self.end_date
        }

    @classmethod
    def from_dict(cls, data):
        if not data:
            return None
        return cls(data["m_type"], data["start_date"], data["end_date"])


class Member(Person):
    def __init__(self, name: str, phone: str, member_id: str):
        super().__init__(name, phone, member_id)
        self.member_id = member_id
        self.membership = None
        self.trainer = None

    def set_membership(self, membership: Membership):
        self.membership = membership

    def set_trainer(self, trainer):
        self.trainer = trainer

    def to_dict(self):
        return {
            "member_id": self.member_id,
            "name": self.name,
            "phone": self.phone,
            "membership": self.membership.to_dict() if self.membership else None,
            "trainer_id": self.trainer.trainer_id if self.trainer else None
        }


class Trainer(Person):
    def __init__(self, name: str, phone: str, trainer_id: str, specialty: str):
        super().__init__(name, phone, trainer_id)
        self.trainer_id = trainer_id
        self.specialty = specialty
        self.members = []

    def add_member(self, member: Member):
        if member not in self.members:
            self.members.append(member)

    def remove_member(self, member: Member):
        if member in self.members:
            self.members.remove(member)

    def to_dict(self):
        return {
            "trainer_id": self.trainer_id,
            "name": self.name,
            "phone": self.phone,
            "specialty": self.specialty
        }


class Payment:
    def __init__(self, payment_id: str, member: Member, amount: float, date_str: str):
        self.payment_id = payment_id
        self.member = member
        self.amount = amount
        self.date_str = date_str

    def to_dict(self):
        return {
            "payment_id": self.payment_id,
            "member_id": self.member.member_id if self.member else None,
            "amount": self.amount,
            "date_str": self.date_str
        }


class Attendance:
    def __init__(self, member: Member, date_str: str):
        self.member = member
        self.date_str = date_str

    def to_dict(self):
        return {
            "member_id": self.member.member_id if self.member else None,
            "date_str": self.date_str
        }


# ==========================================
# 2. GYM SYSTEM WITH FILE HANDLING METHODS
# ==========================================

class Gym:
    def __init__(self, name: str, filename: str = "gym_data.json"):
        self.name = name
        self.filename = filename
        self.members = []
        self.trainers = []
        self.payments = []
        self.attendances = []
        self.load_from_file()

    def check_member_validity(self, member_id: str):
        search_id = str(member_id).strip()
        for m in self.members:
            m_id = getattr(m, 'member_id', getattr(m, 'person_id', None))
            if str(m_id).strip() == search_id:
                return m
        return None

    def check_trainer_validity(self, trainer_id: str):
        search_id = str(trainer_id).strip()
        for t in self.trainers:
            t_id = getattr(t, 'trainer_id', getattr(t, 'person_id', None))
            if str(t_id).strip() == search_id:
                return t
        return None

    def delete_member(self, member_id: str):
        member = self.check_member_validity(member_id)
        if not member:
            return False, f"Member ID '{member_id}' not found."

        if getattr(member, 'trainer', None):
            member.trainer.remove_member(member)

        self.members.remove(member)
        self.save_to_file()
        return True, f"Member '{member.name}' (ID: {member_id}) successfully deleted!"

    # --- FILE HANDLING (SAVE / LOAD) ---

    def save_to_file(self):
        data = {
            "gym_name": self.name,
            "members": [m.to_dict() for m in self.members],
            "trainers": [t.to_dict() for t in self.trainers],
            "payments": [p.to_dict() for p in self.payments],
            "attendances": [a.to_dict() for a in self.attendances]
        }
        with open(self.filename, 'w') as f:
            json.dump(data, f, indent=4)
        return f"Data successfully saved to '{self.filename}'"

    def load_from_file(self):
        if not os.path.exists(self.filename):
            return f"No existing file '{self.filename}' found. Starting fresh."

        try:
            with open(self.filename, 'r') as f:
                data = json.load(f)

            self.members = []
            self.trainers = []
            self.payments = []
            self.attendances = []

            # 1. Load Trainers
            for t_data in data.get("trainers", []):
                trainer = Trainer(t_data["name"], t_data["phone"], t_data["trainer_id"], t_data["specialty"])
                self.trainers.append(trainer)

            # 2. Load Members & Links
            for m_data in data.get("members", []):
                member = Member(m_data["name"], m_data["phone"], m_data["member_id"])

                # Restore Membership
                if m_data.get("membership"):
                    member.set_membership(Membership.from_dict(m_data["membership"]))

                # Restore Trainer Assignment
                t_id = m_data.get("trainer_id")
                if t_id:
                    trainer = self.check_trainer_validity(t_id)
                    if trainer:
                        member.set_trainer(trainer)
                        trainer.add_member(member)

                self.members.append(member)

            # 3. Load Payments
            for p_data in data.get("payments", []):
                member = self.check_member_validity(p_data.get("member_id"))
                if member:
                    payment = Payment(p_data["payment_id"], member, p_data["amount"], p_data["date_str"])
                    self.payments.append(payment)

            # 4. Load Attendances
            for a_data in data.get("attendances", []):
                member = self.check_member_validity(a_data.get("member_id"))
                if member:
                    attendance = Attendance(member, a_data["date_str"])
                    self.attendances.append(attendance)

            return f"Data loaded successfully from '{self.filename}'"
        except Exception as e:
            return f"Error loading file: {e}"


# Initialize Gym Object
gym = Gym("Guindy's Fitness Gym")


# ==========================================
# 3. GRADIO GUI INTERFACE & WRAPPERS
# ==========================================

import gradio as gr

def get_person_phone(obj):
    for attr in ['phone', 'phone_number', '_phone', 'mobile']:
        val = getattr(obj, attr, None)
        if val is not None:
            return val
    return "N/A"

def get_trainer_member_count(trainer_obj):
    for attr in ['members', 'assigned_members', 'client_list', 'member_list', '_members']:
        val = getattr(trainer_obj, attr, None)
        if isinstance(val, (list, set, tuple)):
            return len(val)
    return 0

def get_members_table(search_query=""):
    data = []
    query = str(search_query).strip().lower()

    for m in gym.members:
        m_id = str(getattr(m, 'member_id', getattr(m, 'person_id', 'N/A')))
        m_name = getattr(m, 'name', 'N/A')

        if query and not (query in m_id.lower() or m_name.lower().startswith(query)):
            continue

        ms_obj = getattr(m, 'membership', None)
        if ms_obj and hasattr(ms_obj, 'is_active'):
            ms_status = "Active" if ms_obj.is_active() else "Expired"
        elif ms_obj:
            ms_status = "Active"
        else:
            ms_status = "Inactive/None"

        trainer_obj = getattr(m, 'trainer', None)
        trainer_name = trainer_obj.name if trainer_obj else "None Assigned"

        data.append([
            m_id,
            m_name,
            get_person_phone(m),
            ms_status,
            trainer_name
        ])
    return data

def get_trainers_table():
    data = []
    for t in gym.trainers:
        t_id = getattr(t, 'trainer_id', getattr(t, 'person_id', 'N/A'))
        phone_num = get_person_phone(t)
        specialty = getattr(t, 'specialty', getattr(t, 'spec', 'N/A'))
        assigned_count = get_trainer_member_count(t)

        data.append([
            t_id,
            getattr(t, 'name', 'N/A'),
            phone_num,
            specialty,
            assigned_count
        ])
    return data

def get_dashboard_html():
    total_m = len(gym.members)
    active_m = 0
    for m in gym.members:
        ms = getattr(m, 'membership', None)
        if ms and hasattr(ms, 'is_active') and ms.is_active():
            active_m += 1
        elif ms:
            active_m += 1

    expired_m = total_m - active_m
    total_t = len(gym.trainers)
    total_rev = sum(getattr(p, 'amount', 0) for p in gym.payments)
    total_att = len(gym.attendances)

    return f"""
    <div style="display: grid; grid-template-columns: repeat(3, 1fr); gap: 15px; margin-bottom: 20px;">
        <div style="background-color: #f0f4f8; padding: 15px; border-radius: 10px; border-left: 5px solid #2b6cb0;">
            <h4 style="margin: 0; color: #4a5568;">Total Members</h4>
            <h2 style="margin: 5px 0 0 0; color: #2b6cb0;">{total_m}</h2>
        </div>
        <div style="background-color: #f0fff4; padding: 15px; border-radius: 10px; border-left: 5px solid #38a169;">
            <h4 style="margin: 0; color: #4a5568;">Active Memberships</h4>
            <h2 style="margin: 5px 0 0 0; color: #38a169;">{active_m}</h2>
        </div>
        <div style="background-color: #fff5f5; padding: 15px; border-radius: 10px; border-left: 5px solid #e53e3e;">
            <h4 style="margin: 0; color: #4a5568;">Expired / Inactive</h4>
            <h2 style="margin: 5px 0 0 0; color: #e53e3e;">{expired_m}</h2>
        </div>
        <div style="background-color: #faf5ff; padding: 15px; border-radius: 10px; border-left: 5px solid #805ad5;">
            <h4 style="margin: 0; color: #4a5568;">Total Trainers</h4>
            <h2 style="margin: 5px 0 0 0; color: #805ad5;">{total_t}</h2>
        </div>
        <div style="background-color: #feebc8; padding: 15px; border-radius: 10px; border-left: 5px solid #dd6b20;">
            <h4 style="margin: 0; color: #4a5568;">Total Revenue</h4>
            <h2 style="margin: 5px 0 0 0; color: #dd6b20;">${total_rev:.2f}</h2>
        </div>
        <div style="background-color: #ebf8ff; padding: 15px; border-radius: 10px; border-left: 5px solid #3182ce;">
            <h4 style="margin: 0; color: #4a5568;">Check-in Logs</h4>
            <h2 style="margin: 5px 0 0 0; color: #3182ce;">{total_att}</h2>
        </div>
    </div>
    """

def gui_add_member(m_id, name, phone):
    if not m_id or not name or not phone:
        return "⚠️ All fields are required!", get_members_table()
    if gym.check_member_validity(m_id):
        return f"❌ Error: Member ID '{m_id}' already exists!", get_members_table()
    try:
        new_m = Member(name, phone, m_id)
        gym.members.append(new_m)
        gym.save_to_file()
        return f"✅ Member '{name}' added and saved successfully!", get_members_table()
    except Exception as e:
        return f"❌ Error: {e}", get_members_table()

def gui_delete_member(m_id):
    if not m_id:
        return "⚠️ Please enter a Member ID to delete.", get_members_table(), get_trainers_table()

    success, msg = gym.delete_member(m_id)
    status_icon = "✅" if success else "❌"
    return f"{status_icon} {msg}", get_members_table(), get_trainers_table()

def gui_add_trainer(t_id, name, phone, spec):
    if not t_id or not name or not phone or not spec:
        return "⚠️ All fields are required!", get_trainers_table()
    if gym.check_trainer_validity(t_id):
        return f"❌ Error: Trainer ID '{t_id}' already exists!", get_trainers_table()
    try:
        new_t = Trainer(name, phone, t_id, spec)
        gym.trainers.append(new_t)
        gym.save_to_file()
        return f"✅ Trainer '{name}' added and saved successfully!", get_trainers_table()
    except Exception as e:
        return f"❌ Error: {e}", get_trainers_table()

def gui_assign_membership(m_id, m_type, start_d, end_d):
    member = gym.check_member_validity(m_id)
    if not member:
        return f"❌ Error: Member ID '{m_id}' not found!", get_members_table()
    try:
        ms = Membership(m_type, str(start_d), str(end_d))
        member.set_membership(ms)
        gym.save_to_file()
        return f"✅ Membership '{m_type}' assigned and saved!", get_members_table()
    except Exception as e:
        return f"❌ Error assigning membership: {e}", get_members_table()

def gui_assign_trainer(m_id, t_id):
    member = gym.check_member_validity(m_id)
    trainer = gym.check_trainer_validity(t_id)
    if not member:
        return f"❌ Error: Member ID '{m_id}' not found!", get_members_table(), get_trainers_table()
    if not trainer:
        return f"❌ Error: Trainer ID '{t_id}' not found!", get_members_table(), get_trainers_table()
    try:
        member.set_trainer(trainer)
        trainer.add_member(member)
        gym.save_to_file()
        return f"✅ Trainer '{trainer.name}' assigned to Member '{member.name}' and saved!", get_members_table(), get_trainers_table()
    except Exception as e:
        return f"❌ Error: {e}", get_members_table(), get_trainers_table()

def gui_record_payment(p_id, m_id, amount, p_date):
    member = gym.check_member_validity(m_id)
    if not member:
        return f"❌ Error: Member ID '{m_id}' not found!"
    try:
        p = Payment(p_id, member, float(amount), str(p_date))
        gym.payments.append(p)
        gym.save_to_file()
        return f"✅ Payment of ${float(amount):.2f} recorded and saved!"
    except Exception as e:
        return f"❌ Error recording payment: {e}"

def gui_record_attendance(m_id, a_date):
    member = gym.check_member_validity(m_id)
    if not member:
        return f"❌ Error: Member ID '{m_id}' not found!"
    try:
        att = Attendance(member, str(a_date))
        gym.attendances.append(att)
        gym.save_to_file()
        return f"✅ Attendance logged and saved for '{member.name}'!"
    except Exception as e:
        return f"❌ Attendance Error: {e}"

def gui_manual_save():
    return gym.save_to_file()

def gui_manual_load():
    msg = gym.load_from_file()
    return msg, get_members_table(), get_trainers_table(), get_dashboard_html()


# --- LAYOUT DESIGN ---

custom_theme = gr.themes.Soft(primary_hue="blue", neutral_hue="slate")

with gr.Blocks(theme=custom_theme, title="Gym Management System") as demo:
    gr.Markdown(f"# 🏋️ {gym.name} Management Dashboard")

    # Define DataFrames at a higher scope to ensure they are available
    members_df = gr.Dataframe(
        headers=["ID", "Name", "Phone", "Membership Status", "Trainer"],
        value=get_members_table,
        interactive=False
    )
    trainers_df = gr.Dataframe(
        headers=["ID", "Name", "Phone", "Specialty", "Assigned Members Count"],
        value=get_trainers_table,
        interactive=False
    )

    with gr.Tabs():
        with gr.TabItem("📊 Overview"):
            dash_cards = gr.HTML(get_dashboard_html)
            with gr.Row():
                btn_refresh = gr.Button("🔄 Refresh Overview Stats", variant="secondary")
                btn_save_file = gr.Button("💾 Save Data to File", variant="primary")
                btn_load_file = gr.Button("📁 Load Data from File", variant="secondary")

            file_status = gr.Textbox(label="File Operations Status", interactive=False)

            btn_refresh.click(get_dashboard_html, outputs=dash_cards)
            btn_save_file.click(gui_manual_save, outputs=file_status)
            btn_load_file.click(gui_manual_load, outputs=[file_status, members_df, trainers_df, dash_cards])

        with gr.TabItem("👥 Members"):
            with gr.Row():
                with gr.Column():
                    gr.Markdown("### ➕ Register New Member")
                    m_id = gr.Textbox(label="Member ID", placeholder="e.g. 101")
                    m_name = gr.Textbox(label="Full Name", placeholder="e.g. John Doe")
                    m_phone = gr.Textbox(label="Phone Number", placeholder="e.g. 0123456789")
                    btn_add_m = gr.Button("Add Member", variant="primary")
                    m_status = gr.Textbox(label="Status / Result", interactive=False)

                with gr.Column():
                    gr.Markdown("### 🗑️ Delete Member")
                    del_m_id = gr.Textbox(label="Member ID to Delete", placeholder="e.g. 101")
                    btn_del_m = gr.Button("Delete Member", variant="stop")
                    del_status = gr.Textbox(label="Delete Result", interactive=False)

            gr.Markdown("---")
            gr.Markdown("### 📋 Member Directory")
            search_bar = gr.Textbox(label="🔍 Search Members (by Name prefix or ID)", placeholder="Type to filter...")

            # members_df = gr.Dataframe(
            #     headers=["ID", "Name", "Phone", "Membership Status", "Trainer"],
            #     value=get_members_table,
            #     interactive=False
            # )

            search_bar.change(get_members_table, inputs=search_bar, outputs=members_df)
            btn_add_m.click(gui_add_member, inputs=[m_id, m_name, m_phone], outputs=[m_status, members_df])
            btn_del_m.click(gui_delete_member, inputs=del_m_id, outputs=[del_status, members_df, trainers_df])

        with gr.TabItem("🧘 Trainers"):
            gr.Markdown("### ➕ Register New Trainer")
            with gr.Row():
                t_id = gr.Textbox(label="Trainer ID", placeholder="e.g. 201")
                t_name = gr.Textbox(label="Full Name", placeholder="e.g. Coach Alex")
                t_phone = gr.Textbox(label="Phone Number", placeholder="e.g. 0987654321")
                t_spec = gr.Textbox(label="Specialty", placeholder="e.g. Fitness / Bodybuilding")
            btn_add_t = gr.Button("Add Trainer", variant="primary")
            t_status = gr.Textbox(label="Status / Result", interactive=False)

            gr.Markdown("### 📋 Trainer Directory")
            # trainers_df = gr.Dataframe(
            #     headers=["ID", "Name", "Phone", "Specialty", "Assigned Members Count"],
            #     value=get_trainers_table,
            #     interactive=False
            # )
            btn_add_t.click(gui_add_trainer, inputs=[t_id, t_name, t_phone, t_spec], outputs=[t_status, trainers_df])

        with gr.TabItem("💳 Memberships & Trainers"):
            with gr.Row():
                with gr.Column():
                    gr.Markdown("### 📝 Assign Membership")
                    ms_mid = gr.Textbox(label="Member ID", placeholder="e.g. 101")
                    ms_type = gr.Dropdown(["Weekly", "Monthly", "Yearly"], label="Membership Type", value="Monthly")
                    ms_start = gr.Textbox(label="Start Date (YYYY-MM-DD)", value="2026-01-01")
                    ms_end = gr.Textbox(label="End Date (YYYY-MM-DD)", value="2026-12-31")
                    btn_assign_ms = gr.Button("Assign Membership", variant="primary")
                    ms_status = gr.Textbox(label="Result", interactive=False)
                    btn_assign_ms.click(gui_assign_membership, inputs=[ms_mid, ms_type, ms_start, ms_end], outputs=[ms_status, members_df])

                with gr.Column():
                    gr.Markdown("### 🤝 Assign Trainer to Member")
                    tr_mid = gr.Textbox(label="Member ID", placeholder="e.g. 101")
                    tr_tid = gr.Textbox(label="Trainer ID", placeholder="e.g. 201")
                    btn_assign_tr = gr.Button("Link Trainer & Member", variant="primary")
                    tr_status = gr.Textbox(label="Result", interactive=False)
                    btn_assign_tr.click(gui_assign_trainer, inputs=[tr_mid, tr_tid], outputs=[tr_status, members_df, trainers_df])

        with gr.TabItem("💰 Payments"):
            gr.Markdown("### 💵 Record Payment")
            with gr.Row():
                p_id = gr.Textbox(label="Payment ID", placeholder="e.g. PAY-1")
                p_mid = gr.Textbox(label="Member ID", placeholder="e.g. 101")
                p_amt = gr.Number(label="Amount ($", value=50.0)
                p_date = gr.Textbox(label="Payment Date (YYYY-MM-DD)", value="2026-09-30")
            btn_pay = gr.Button("Record Payment", variant="primary")
            p_status = gr.Textbox(label="Status / Result", interactive=False)
            btn_pay.click(gui_record_payment, inputs=[p_id, p_mid, p_amt, p_date], outputs=p_status)

        with gr.TabItem("📅 Check-ins"):
            gr.Markdown("### 📌 Log Member Check-in")
            with gr.Row():
                a_mid = gr.Textbox(label="Member ID", placeholder="e.g. 101")
                a_date = gr.Textbox(label="Check-in Date (YYYY-MM-DD)", value="2026-09-30")
            btn_att = gr.Button("Log Attendance", variant="primary")
            a_status = gr.Textbox(label="Status / Result", interactive=False)
            btn_att.click(gui_record_attendance, inputs=[a_mid, a_date], outputs=a_status)

demo.launch(inline=True)
```
